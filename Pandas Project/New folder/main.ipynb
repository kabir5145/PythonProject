```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("netflix_titles.csv")
print(df.head())
print(df.shape)
print(df.columns)
```

```python
# total movies and TV shows
total_movies = df[df["type"] == "Movie"].shape[0]
total_shows = df[df["type"] == "TV Show"].shape[0]

print(f"Total Movies: {total_movies}")
print(f"Total TV Shows: {total_shows}")

plt.bar(["Movies", "TV Shows"], [total_movies, total_shows], color=["blue", "red"])
plt.title("Total Movies vs TV Shows on Netflix",color="green")
plt.xlabel("Content Type")
plt.ylabel("Count")
plt.show()
```

```python
# Top 10 rated movies and TV shows
top_movies = df[df["type"] == "Movie"].sort_values(by="rating", ascending=False).head(10)
top_shows = df[df["type"] == "TV Show"].sort_values(by="rating", ascending=False).head(10)


print("Top 10 Movies:")
print(top_movies[["title", "rating"]])

print("\nTop 10 TV Shows:")
print(top_shows[["title", "rating"]])
```

```python
# Most common genres
most_common_genres = df["listed_in"].str.split(", ").explode().value_counts().head(10)
print("\nMost Common Genres:")
print(most_common_genres)
```

```python
# Release year distribution
plt.hist(df["release_year"].dropna(), bins=20, color="purple", edgecolor="black")
plt.title("Distribution of Release Years",color="green")
plt.xlabel("Release Year")
plt.ylabel("Frequency")
plt.show()
```

```python
# Most common rated 
most_common_ratings = df["rating"].value_counts().head(10)
print("\nMost Common Ratings:")
print(most_common_ratings)
```

```python
# Top 10 countries with most content
top_countries = df["country"].str.split(", ").explode().value_counts().head(10)
print("\nTop 10 Countries with Most Content:")
print(top_countries)
```

```python
# Movies and TV shows of India
indian_content = df[df["country"].str.contains("India", na=False)]
print("\nMovies and TV Shows from India:")
print(indian_content[["title", "type", "release_year"]].head(10))
```

```python
# Longest duration movies and TV shows
longest_movie = df[df["type"]== "Movie"].sort_values(by= "duration", ascending=False).head(1)
longest_show = df[df["type"]== "TV Show"].sort_values(by= "duration", ascending=False).head(1)

print("\nLongest Movie:")
print(longest_movie[["title", "duration"]])

print("\nLongest TV Show:")
print(longest_show[["title", "duration"]])

# Sortest duration movies and TV shows
shortest_movie = df[df["type"]== "Movie"].sort_values(by= "duration", ascending=True).head(1)
shortest_show = df[df["type"]== "TV Show"].sort_values(by= "duration", ascending=True).head(1)

print("\nShortest Movie:")
print(shortest_movie[["title", "duration"]])

print("\nShortest TV Show:")
print(shortest_show[["title", "duration"]])
```
