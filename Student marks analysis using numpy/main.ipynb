```python
import numpy as np

# Columns: [Math, Science, English, History, Computer Science]
student_marks = np.array([
    [85, 78, 92, 74, 88],  # Student 1
    [76, 85, 89, 80, 90],  # Student 2
    [90, 88, 94, 86, 91],  # Student 3
    [65, 70, 72, 68, 75],  # Student 4
    [88, 92, 85, 90, 94],  # Student 5
    [70, 68, 75, 72, 78],  # Student 6
    [95, 94, 96, 92, 98],  # Student 7
    [82, 80, 78, 85, 84],  # Student 8
    [60, 65, 70, 62, 68],  # Student 9
    [78, 75, 80, 77, 82]   # Student 10
])

student_names = np.array([
    "Alice", "Bob", "Charlie", "David", "Eva",
    "Frank", "Grace", "Hannah", "Ian", "Jack"
])

print("Student Marks Array:\n", student_marks)
```

```python
# Function to calculate percentage for each student.
def cal_percentage(scores):
    total = np.sum(scores)
    percentage = (total / 500) * 100
    return percentage

print("Student Scores and Percentages:")
for i, student in enumerate(student_marks):
    print(f"Student {i+1} Scores: {student}, Percentage: {cal_percentage(student):.2f}%")
```

```python
# Function to calculate average marks for each subject.
def cal_average(marks):
    return np.mean(marks, axis=0)

average_marks = cal_average(student_marks)
print("\nAverage Marks for Each Subject:")
subjects = ["Math", "Science", "English", "History", "Computer Science"]

for subject, avg in zip(subjects, average_marks):
    print(f"{subject}: {avg:.2f}")
```

```python
#Function to calculate Highest and lowest marks for each subject.
def cal_high_low(marks):
    highest = np.max(marks, axis =0)
    lowest = np.min(marks, axis =0)
    return highest, lowest

highest_marks, lowest_marks = cal_high_low(student_marks)
print("\nHighest and Lowest Marks for Each Subject:")

for subject , high, low in zip(subjects, highest_marks, lowest_marks):
    print(f"{subject}: Highest: {high}, Lowest: {low}")
```

```python
#Function to find the topper in the class based on total marks.
def find_topper(marks):
    total_marks = np.sum(marks, axis=1)
    topper_index = np.argmax(total_marks)
    return topper_index, total_marks[topper_index]

topper_index, topper_marks = find_topper(student_marks)
print(f"\nTopper: Student {topper_index + 1} with Total Marks: {topper_marks}")
print(f"Topper's Name: {student_names[topper_index]}")
```
